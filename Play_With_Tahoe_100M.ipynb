```python
from datasets import load_dataset
from scipy.sparse import csr_matrix
import anndata
import pandas as pd
import pubchempy as pcp
```

```python
def create_anndata_from_generator(generator, gene_vocab, sample_size=None):
    sorted_vocab_items = sorted(gene_vocab.items())
    token_ids, gene_names = zip(*sorted_vocab_items)
    token_id_to_col_idx = {token_id: idx for idx, token_id in enumerate(token_ids)}

    data, indices, indptr = [], [], [0]
    obs_data = []

    for i, cell in enumerate(generator):
        if sample_size is not None and i >= sample_size:
            break
        genes = cell['genes']
        expressions = cell['expressions']
        if expressions[0] < 0:
            genes = genes[1:]
            expressions = expressions[1:]

        col_indices = [token_id_to_col_idx[gene] for gene in genes if gene in token_id_to_col_idx]
        valid_expressions = [expr for gene, expr in zip(genes, expressions) if gene in token_id_to_col_idx]

        data.extend(valid_expressions)
        indices.extend(col_indices)
        indptr.append(len(data))

        obs_entry = {k: v for k, v in cell.items() if k not in ['genes', 'expressions']}
        obs_data.append(obs_entry)

    expr_matrix = csr_matrix((data, indices, indptr), shape=(len(indptr) - 1, len(gene_names)))
    obs_df = pd.DataFrame(obs_data)

    adata = anndata.AnnData(X=expr_matrix, obs=obs_df)
    adata.var.index = pd.Index(gene_names, name='ensembl_id')

    return adata
```

```python
import pandas as pd
import gzip

def parse_geo_metadata(file_path):
    # Initialize dictionary to store data
    metadata = {
        "geo_accession": [],
        "sample_title": [],
        "sex": [],
        "condition": []
    }

    # Logic to handle different file encodings/types if needed
    try:
        f = open(file_path, 'r')
    except:
        f = gzip.open(file_path, 'rt')

    with f:
        for line in f:
            # Clean up line
            line = line.strip().replace('"', '')

            # Extract GSM Accessions
            if line.startswith('!Sample_geo_accession'):
                # Skip the first element which is the header "!Sample_geo_accession"
                metadata["geo_accession"] = line.split('\t')[1:]

            # Extract Sample Titles (e.g. SCP3412)
            elif line.startswith('!Sample_title'):
                metadata["sample_title"] = line.split('\t')[1:]

            # Extract Characteristics (Sex and Condition)
            elif line.startswith('!Sample_characteristics_ch1'):
                values = line.split('\t')[1:]

                # Check what kind of characteristic this row is
                if "genotype:" in values[0]:
                    # Clean the string to keep only 'male' or 'female'
                    metadata["sex"] = [x.replace('genotype: ', '') for x in values]

                elif "treatment:" in values[0]:
                    # Clean the string to keep only 'Preeclampsia' or 'Control'
                    metadata["condition"] = [x.replace('treatment: ', '') for x in values]

    # Create DataFrame
    df = pd.DataFrame(metadata)

    return df
```

```python

# Usage
# Save your file as 'GSE306864_series_matrix.txt' in your folder
metadata_df = parse_geo_metadata('GSE306864_series_matrix.txt')

# Inspect the cleaned data
print(metadata_df.head())

# Save to CSV for your pipeline
metadata_df.to_csv("GSE306864_metadata_cleaned.csv", index=False)
```
